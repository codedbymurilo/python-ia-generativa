```python
# PROCESSAMENTO DE LINGUAGEM NATURAL
# Objeivo
# Usar spaCy para preparar/limpar textos em português
# e NLTK para fazer uma análise de sentimento

# Importando as bibliotecas
import spacy
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python
# PREPARAÇÃO DO AMBIENTE
!python -m spacy download pt_core_news_sm
nltk.download('vader_lexicon')
nlp = spacy.load("pt_core_news_sm")
termometro_nltk = SentimentIntensityAnalyzer()
```

```python
# AJUSTE DO DICIONÁRIO DE SENTIMENTOS
pesos_extras = {
    # Palavras positivas
    "maravilhoso": 2.0,
    "adorar": 1.5,
    "incrível": 2.0,
    "incrivel": 2.0,
    "facil": 1.0,
    "fácil": 1.0,
    "atencioso": 1.0,
    # Palavras negativas
    "péssimo": -2.0,
    "pessimo": -2.0,
    "ruim": -1.5,
    "confuso": -1.5,
    "caro": -1.0,
    "demorar": -1.0,
    # Palavras importantes em sentimento
    "não": -0.8
}
termometro_nltk.lexicon.update(pesos_extras)
```

```python
# DADOS DE EXEMPLO
frases_clientes = [
    "O curso do Senai é maravilhoso, os professores são muito atenciosos!",
    "Péssimo atendimento. O suporte demorou demais para responder.",
    "A aula de hoje foi normal, nada de muito especial.",
    "EU ADOREI o conteúdo de IA! É muito fácil de entender.",
    "Não gostei do material, achei confuso e muito caro.",
    "O evento em São Paulo foi incrível, mas o café estava ruim."
]
```

```python
# EXECUÇÃO INTEGRADA: spaCy + NLTK
print(f"{'FEEDBACK ORIGINAL':<70} | {'NOTA':<6} | {'SENTIMENTO'}")
print("-" * 90)

for texto in frases_clientes:
    texto_minusculo = texto.lower()
    doc = nlp(texto_minusculo)
    palavras_uteis = []
    for token in doc:
        if token.is_punct:
            continue
        if token.is_stop and token.text not in ["não","nunca","jamais"]:
            continue
        palavras_uteis.append(token.lemma_)
    #print("Palavras úteis:", palavras_uteis)
    frase_limpa = " ".join(palavras_uteis)

    # Análise de sentimento
    nota = termometro_nltk.polarity_scores(frase_limpa)['compound']

    # Classificação do sentimento 🫵😂
    # nota >= 0.1 -> sentimento positivo
    # nota <= -0.1 -> sentimento negativo
    # nota = 0 -> sentimento neutro

    if nota >= 0.1:
        resultado = "😁 POSITIVO"
    elif nota <= -0.1:
        resultado = "😡 NEGATIVO"
    else:
        resultado = "😐 NEUTRO"

    print(f"{texto:<70} {nota:<6.1f} | {resultado}\n")
```
