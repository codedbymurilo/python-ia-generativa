```python
# CÉLULA 1: Preparação dos dados
import pandas as pd

# Os dados históricos da nossa loja
dados_sorvetes = {
    'temperatura_c': [20, 22, 25, 28, 30, 35],
    'sorvetes_vendidos': [50, 65, 80, 100, 120, 180]
}

tabela_sorvetes = pd.DataFrame(dados_sorvetes)
print("O nosso histórico de vendas:")
display(tabela_sorvetes)
```

```python
# CÉLULA 2: Visualizar a correlação
tabela_sorvetes.plot(
    kind='scatter', x='temperatura_c', y='sorvetes_vendidos',
    color='purple', title='Temperatura vs Vendas'
)
```

```python
# CÉLULA 3: O treino e a predição
from sklearn.linear_model import LinearRegression

# 1. Separar a pergunta (X) da resposta (y)
X = tabela_sorvetes[['temperatura_c']]
y = tabela_sorvetes['sorvetes_vendidos']

# 2. Treinar o Cérebro da IA
modelo_sorvetes = LinearRegression()
modelo_sorvetes.fit(X,y)

# 3. Predição: queremos saber o estoque necessário para 38 graus
temperatura = pd.DataFrame({'temperatura_c':[38]})
estimativa = modelo_sorvetes.predict(temperatura)

print(f"ALERTA DE ESTOQUE: Estimativa de vendas: {int(estimativa[0])}")
```
