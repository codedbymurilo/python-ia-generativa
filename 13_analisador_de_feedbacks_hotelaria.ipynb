```python
# PROCESSAMENTO DE LINGUAGEM NATURAL
# Objeivo
# Usar spaCy para preparar/limpar textos em português
# e NLTK para fazer uma análise de sentimento

# Importando as bibliotecas
import spacy
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python
# PREPARAÇÃO DO AMBIENTE
!python -m spacy download pt_core_news_sm
nltk.download('vader_lexicon')
nlp = spacy.load("pt_core_news_sm")
termometro_nltk = SentimentIntensityAnalyzer()
```

```python
# AJUSTE DO DICIONÁRIO DE SENTIMENTOS
pesos_extras = {
    # Termos Positivos (Hotelaria)
    "aconchegante": 1.8,
    "impecável": 2.0,
    "impecavel": 2.0,
    "limpo": 1.2,
    "limpeza": 1.2,
    "solícito": 1.5,
    "solicito": 1.5,
    "saboroso": 1.5,
    "top": 1.5,

    # Termos Negativos (Hotelaria)
    "sujo": -2.0,
    "sujeira": -2.0,
    "barulho": -1.2,
    "barulhento": -1.5,
    "vazar": -1.5,
    "vazamento": -1.5,
    "mofado": -1.8,
    "mofo": -1.8,
    "demorar": -1.5,
    "péssimo": -2.0,
    "pessimo": -2.0,
    # Modificadores de Sentimento
    "não": -0.8
}

termometro_nltk.lexicon.update(pesos_extras)
```

```python
# DADOS DE EXEMPLO
frases_clientes = [
    "O quarto estava impecável e o café da manhã era top demais!",
    "Péssima experiência, o chuveiro vazou a noite toda e o quarto tinha cheiro de mofo.",
    "A cama é aconchegante, mas a recepção demorou um século pra fazer meu check-in.",
    "Atendimento nota dez, os funcionários foram super solícitos e o quarto estava muito limpo.",
    "Não curti a estadia, muito barulho no corredor e o ar-condicionado estava ruim."
]
```

```python
# EXECUÇÃO INTEGRADA: spaCy + NLTK
print(f"{'FEEDBACK ORIGINAL':<95} | {'NOTA':<6} | {'SENTIMENTO'}")
print("-" * 122)

for texto in frases_clientes:
    texto_minusculo = texto.lower()
    doc = nlp(texto_minusculo)
    palavras_uteis = []
    for token in doc:
        if token.is_punct:
            continue
        if token.is_stop and token.text not in ["não","nunca","jamais"]:
            continue
        palavras_uteis.append(token.lemma_)

    #print("Palavras úteis:", palavras_uteis)
    frase_limpa = " ".join(palavras_uteis)

    # Análise de sentimento
    nota = termometro_nltk.polarity_scores(frase_limpa)['compound']

    # Classificação do sentimento
    if nota >= 0.2:
        resultado = "😁 POSITIVO"
    elif nota <= -0.1:
        resultado = "😡 NEGATIVO"
    else:
        resultado = "😐 NEUTRO"

    # Print com alinhamento de 95 caracteres na coluna do texto
    print(f"{texto:<95} | {nota:<6.1f} | {resultado}\n")

```
