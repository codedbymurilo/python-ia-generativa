```python
# Dados para a dinâmica
import pandas as pd

# Histórico dos meses anteriores
dados_marketing = {
    'Investimento_Anuncios': [1000, 2000, 3000, 4000, 5000],
    'Faturamento': [15000, 25000, 36000, 48000, 58000]
}

tabela_marketing = pd.DataFrame(dados_marketing)
print("O nosso histórico de faturamento:")
display(tabela_marketing)
```

```python
# Gráfico de dispersão
tabela_marketing.plot(
    kind='bar', x='Investimento_Anuncios', y='Faturamento',
    color='lightseagreen', title='Investimento no Anuncios vs Faturamento'
)
```

```python
# Preparação e Treino
from sklearn.linear_model import LinearRegression

# 1. Separar a pergunta (X) da resposta (y)
X = tabela_marketing[['Investimento_Anuncios']]
y = tabela_marketing['Faturamento']

# 2. Treinar o Cérebro da IA
modelo_marketing = LinearRegression()
modelo_marketing.fit(X,y)

# Input
investimento = float(input("Informe o investimento: "))

# 3. Predição: queremos saber o estoque necessário para 38 graus
investimento_anuncios = pd.DataFrame({'Investimento_Anuncios':[investimento]})
faturamento_previsto = modelo_marketing.predict(investimento_anuncios)

print(f"Estimativa de faturamento: {faturamento_previsto[0]:,.2f}")
```
