```python
import pandas as pd
import joblib
import tensorflow as tf

# ============================================================
# 1. Carregar o modelo treinado e os normalizadores
# ============================================================

modelo_limite = tf.keras.models.load_model('modelo_limite_credito.keras')

normalizador_x = joblib.load('normalizador_x.pkl')
normalizador_y = joblib.load('normalizador_y.pkl')


# ============================================================
# 2. Criar os dados de um novo cliente
# ============================================================

cliente_novo = pd.DataFrame({
    'Idade': [27],
    'Salario': [4500],
    'Score_Credito': [1000]
})


# ============================================================
# 3. Normalizar os dados do cliente
# ============================================================

cliente_norm = normalizador_x.transform(cliente_novo)


# ============================================================
# 4. Fazer a previsão
# ============================================================

limite_previsto_norm = modelo_limite.predict(cliente_norm)


# ============================================================
# 5. Converter o resultado para Reais
# ============================================================

limite_real = normalizador_y.inverse_transform(limite_previsto_norm)

print(f"O limite de crédito sugerido é: R$ {limite_real[0][0]:.2f}")
```
