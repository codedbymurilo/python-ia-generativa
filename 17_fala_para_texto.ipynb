```python
# RECONHECIMENTO DE FALA COM ARQUIVO .WAV

import speech_recognition as sr

# 1. Inicializar o reconhecedor de voz da IA
reconhecedor = sr.Recognizer()

# 2. Abrir o arquivo de áudio e extrair o texto via IA
arquivos = ['nenhuma.wav', 'reconhecer.wav', 'detectar.wav']
nome_arquivo = arquivos[2]

print(f"Abrindo o arquivo '{nome_arquivo}' e preparando os dados ...")

# 3. Bloco 'with' abre o arquivo de áudio como uma fonte de dados para o Python
with sr.AudioFile(nome_arquivo) as fonte:
    reconhecedor.adjust_for_ambient_noise(fonte, duration=0.5)
    dados_audio = reconhecedor.record(fonte)

print("Enviando o áudio para o processamento na rede neural ...")

try:
    # Chamar o motor de busca em português (pt-BR)
    texto_transcrito = reconhecedor.recognize_google(dados_audio, language='pt-BR')

    # Padroniza o texto para letras maiúsculas para facilitar a comparação lógica
    comando_final = texto_transcrito.upper()
    print(f"Texto transcrito: {comando_final}")

# 4. Tratamento de erros caso o áudio esteja mudo ou incompreensível
except sr.UnknownValueError:
    print("\nNão foi possível decifrar os fonemas.")
    comando_final = "ERRO"
except Exception as e:
    print(f"Falha na conexão ou arquivo corrompido: {e}")
    comando_final = "ERRO"

# 5. Criar a árvore de decisão
if "DETECTAR" in comando_final:
    print("\n[AÇÃO] >>> Comando válido! Disparando Rede Neural YOLO ...")

elif "RECONHECER" in comando_final:
    print("\n[AÇÃO] >>> Comando válido Disparando comparação biométrica facial ...")
else:
    print("\n[ALERTA] >>> Comando não mapeado ou inválido.")
```
